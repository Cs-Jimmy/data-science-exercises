### Now do the following

#### Get the sum of all the values in mat using for loop

```python
sum=0
arr=mat.reshape(-1)
for num in arr:
    sum+=num
print(sum)
```

#### Get the standard deviation of the values in mat (without built in fuction)

```python
arr=mat.reshape(-1)   #flatten the array
mean=arr.sum()/len(arr)
sumofsq=0
for num in arr:
    sumofsq+=((num-mean)**2)
std=(sumofsq/len(arr))**0.5
print(std)

```

##### Get the sum of all the columns in mat and put in nparray

```python
nparray=np.array(mat.sum(axis=0))
print(nparray)
```
